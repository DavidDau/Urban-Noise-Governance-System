# Model Comparison

This notebook compares the performance of the Random Forest baseline model and the Convolutional Neural Network (CNN).

The objective is to evaluate whether deep learning provides a performance improvement over traditional machine learning techniques for urban noise classification.

Outputs generated:
- Model performance table
- Accuracy comparison chart
- Comparison CSV file

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

## Define Model Performance Metrics

The metrics below were obtained from the training and evaluation notebooks.

Random Forest metrics were generated in 04_baseline_models.ipynb.

CNN metrics were generated in 05_cnn_model.ipynb.

```python
comparison_data = {
    "Model": [
        "Random Forest",
        "CNN"
    ],
    "Accuracy": [
        0.82,
        0.9199
    ],
    "Precision": [
        0.81,
        0.9216
    ],
    "Recall": [
        0.80,
        0.9199
    ],
    "F1 Score": [
        0.80,
        0.9197
    ]
}

results_df = pd.DataFrame(comparison_data)

results_df
```

## Save Comparison Results

The performance metrics are saved as a CSV file for reporting and future reference.

```python
OUTPUT_DIR = "/content/drive/MyDrive/UrbanNoiseProject/outputs"

results_df.to_csv(
    f"{OUTPUT_DIR}/model_comparison.csv",
    index=False
)

print("Comparison CSV saved.")
```

## Visualize Model Accuracy

A bar chart is generated to compare classification accuracy across models.

```python
plt.figure(figsize=(8,5))

plt.bar(
    results_df["Model"],
    results_df["Accuracy"]
)

plt.ylabel("Accuracy")
plt.xlabel("Model")
plt.title("Model Accuracy Comparison")

plt.tight_layout()

plt.savefig(
    f"{OUTPUT_DIR}/model_accuracy_comparison.png"
)

plt.show()
```

## Findings

The CNN achieved the highest performance across all evaluation metrics.

Compared with the Random Forest baseline, the CNN demonstrated superior capability in learning complex acoustic patterns from Mel Spectrogram representations of urban noise.

Therefore, the CNN is selected as the final model for deployment in the Urban Noise Governance System.
